```python
df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
num_cols = df.select_dtypes(include="number").columns
cat_cols = df.select_dtypes(include="object").columns

for c in num_cols:
    df[c] = df[c].fillna(df[c].median())
for c in cat_cols:
    df[c] = df[c].fillna(df[c].mode()[0])

print(df.isna().sum())   # all should be 0
```

```python
for c in cat_cols:
    df[c] = df[c].str.strip()
df.head()
```

```python
df.loc[2]    #row selection
```

```python
df[['first_name','age','gender']].head()
```

```python
df.columns = df.columns.str.title()   #capitalising each title of column
```

```python
df.columns = df.columns.str.replace('_',' ').str.title()   #replacing underscores with spaces
```

```python
df.columns = df.columns.str.upper()   #all in uppercase
```

```python
df.iloc[2]    #return 0 index row
```

```python
#missing data
df['AGE']= df['AGE'].fillna(df['AGE'].mean())
```

```python
df['INTERNET ACCESS AT HOME']=df['INTERNET ACCESS AT HOME'].fillna(df['INTERNET ACCESS AT HOME'].mode()[0])
```

```python
df.duplicated().sum()    #removing duplicates
df = df.drop_duplicates().reset_index(drop=True)
```

```python
print(df['DEPARTMENT'].mode())
```

```python
print(df.columns.tolist())
```

```python
#fixing data types
df['GENDER']=df['GENDER'].astype('category')
print(df.dtypes)
```

```python
df.isnull().sum()
```

```python
#feature engineering
df['FULL NAME']=df['FIRST NAME']+' '+df['LAST NAME']
df['AVG SCORE']=df[['MIDTERM SCORE', 'FINAL SCORE','ASSIGNMENTS AVG','QUIZZES AVG']].mean(axis=1)
```

```python
df['STUDY SLEEP RATIO'] = df['STUDY HOURS PER WEEK'] / df['SLEEP HOURS PER NIGHT']
df['SCORE IMPROVEMENT'] = df['FINAL SCORE'] - df['MIDTERM SCORE']
```
